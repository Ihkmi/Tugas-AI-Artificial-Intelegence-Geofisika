# PROMPT 1 — SEISMOLOGI (DIAGRAM WADATI)

## 1. Narasi Permasalahan
Dalam bidang geofisika, salah satu permasalahan yang sering dihadapi oleh seorang geofisikawan adalah mengidentifikasi parameter gempa bumi berdasarkan data waktu tiba gelombang P dan gelombang S. Data tersebut dapat digunakan untuk menentukan waktu terjadinya gempa (*origin time*) serta memperkirakan parameter lain yang berkaitan dengan sumber gempa. Penentuan parameter gempa secara cepat dan presisi merupakan langkah krusial dalam mitigasi bencana seismik serta pemetaan aktivitas tektonik regional. Tanpa penentuan waktu asal yang akurat, analisis lanjutan mengenai kedalaman fokus dan posisi hiposenter gempa dapat mengalami penyimpangan signifikan.

Salah satu metode klasik namun sangat handal untuk menganalisis hubungan antara waktu tiba gelombang P dan S adalah **Diagram Wadati**. Metode ini memanfaatkan perbedaan waktu tiba gelombang S dan P ($T_s - T_p$) dari beberapa stasiun seismik yang merekam kejadian gempa yang sama. Dengan melakukan analisis linier terhadap data tersebut, dapat diperoleh informasi penting mengenai **origin time ($T_0$)** dan rasio kecepatan gelombang **$V_p/V_s$**. Rasio $V_p/V_s$ ini juga mencerminkan sifat elastisitas serta saturasi fluida pada batuan di sepanjang jalur penjalaran gelombang.

---

## 2. Struktur Good Prompt

### R — Role
Anda berperan sebagai seorang **geofisikawan dan seismolog** yang ahli dalam pengolahan data seismik, analisis gelombang P dan S, serta metode Diagram Wadati.

### T — Task
Tugas Anda adalah membantu mengolah data waktu tiba gelombang P dan S dari beberapa stasiun seismik untuk menentukan **origin time ($T_0$)** dan **rasio $V_p/V_s$** menggunakan metode Diagram Wadati.

### C — Context
Data yang digunakan merupakan data waktu tiba gelombang P dan S dari beberapa stasiun seismik yang merekam kejadian gempa yang sama. Analisis dilakukan secara sistematis untuk memperkirakan parameter gempa dan memverifikasi hasilnya berdasarkan prinsip dasar seismologi.

### I — Input
Data masukan berupa tabel yang terdiri dari:
1. Nama/kode stasiun.
2. Waktu tiba gelombang P .
3. Waktu tiba gelombang S .
4. Selisih waktu $T_s - T_p$.

### K — Constraints
1. Gunakan prinsip dasar seismologi dan metode Diagram Wadati.
2. Tampilkan rumus yang digunakan dalam setiap tahap.
3. Gunakan satuan waktu yang konsisten.
4. Jangan mengubah atau membuat data baru.
5. Gunakan bahasa Indonesia yang sistematis.

### O — Output
1. Tabel hasil pengolahan data.
2. Langkah perhitungan dan rumus.
3. Persamaan regresi linear (slope dan intercept).
4. Parameter gempa ($T_0$ dan $V_p/V_s$).
5. Interpretasi dan verifikasi hasil.

### Q — Quality Criteria
Hasil harus konsisten secara perhitungan, sesuai prinsip dasar seismologi, dan mencantumkan satuan yang jelas.

### V — Verification
Lakukan pengecekan ulang terhadap perhitungan $T_s - T_p$, konsistensi satuan, serta kelogisan nilai $T_0$ sebelum waktu tiba $T_p$.

# PROMPT 2 — GEOLISTRIK RESISTIVITAS

## 1. Narasi Permasalahan
Metode geolistrik resistivitas merupakan salah satu metode geofisika aktif yang digunakan untuk mengetahui variasi sifat kelistrikan material bawah permukaan. Prinsip kerjanya dilakukan dengan menginjeksikan arus listrik ke dalam tanah melalui elektroda arus dan mengukur beda potensial yang dihasilkan pada elektroda potensial. Perbedaan respon hambatan listrik ini dipengaruhi oleh ketersediaan fluida, porositas batuan, tingkat salinitas, serta kandungan mineral lempung di bawah permukaan.

Dalam pengolahan data geolistrik, tantangan utama seorang geofisikawan adalah mengidentifikasi dan memetakan persebaran perlapisan tanah, batuan, zona jenuh air (*akuifer*), serta batas struktur geologi bawah permukaan. Oleh karena nilai resistivitas satu jenis batuan dapat saling tumpang-tindih (*overlapping*), diperlukan analisis kurva resistivitas semu dan pemodelan yang cermat agar interpretasi litologi tidak bias dan dapat menggambarkan kondisi geologi yang sebenarnya secara representatif.

---

## 2. Struktur Good Prompt

### R — Role
Anda berperan sebagai seorang **geofisikawan yang ahli dalam metode geolistrik resistivitas dan interpretasi bawah permukaan**.

### T — Task
Tugas Anda adalah menganalisis data resistivitas bawah permukaan dan memberikan interpretasi terhadap kemungkinan jenis material atau perlapisan.

### C — Context
Data diperoleh melalui pengukuran sifat kelistrikan menggunakan konfigurasi elektroda tertentu. Interpretasi harus mempertimbangkan kondisi geologi daerah penelitian.

### I — Input
Data masukan berupa:
1. Jarak/konfigurasi elektroda.
2. Arus listrik ($I$) dan beda potensial ($\Delta V$).
3. Nilai resistivitas semu ($\rho_a$).

### K — Constraints
1. Gunakan prinsip dasar geolistrik.
2. Jangan menentukan jenis batuan secara pasti hanya dari satu nilai resistivitas.
3. Sertakan konteks geologi dan ketidakpastian.

### O — Output
1. Tabel pengolahan data.
2. Perhitungan faktor geometri ($K$) dan resistivitas semu.
3. Identifikasi zona resistivitas
4. Interpretasi geologi dan kesimpulan.

### Q — Quality Criteria
Perhitungan konsisten, menggunakan rumus geometri yang benar, serta interpretasi tidak berlebihan.

### V — Verification
Periksa kembali perkalian faktor geometri, satuan resistivitas ($\Omega\cdot\text{m}$), serta kesesuaian nilai dengan literatur geologi.

# PROMPT 3 — SEISMIK REFRAKSI

## 1. Narasi Permasalahan
Dalam bidang geofisika dangkal (*near-surface geophysics*), metode seismik refraksi merupakan metode yang memanfaatkan gelombang seismik terbiaskan secara kritis di sepanjang batas perlapisan bawah permukaan. Gelombang yang diinisiagikan dari sumber getaran (*shotpoint*) akan merambat menembus medium tanah dan terbias balik menuju deretan geofon yang terpasang di permukaan. Metode ini sangat efektif digunakan dalam eksplorasi rekayasa sipil, penentuan kedalaman batuan dasar (*bedrock*), serta analisis stabilitas lereng.

Permasalahan mendasar yang kerap dijumpai adalah mengolah data waktu tiba gelombang pertama (*first arrival time*) untuk memisahkan antara gelombang langsung (*direct wave*) dan gelombang bias (*refracted wave*). Dari pemisahan tersebut, seorang geofisikawan harus mampu menghitung kecepatan penjalaran gelombang pada tiap lapisan ($V_1, V_2, \dots$) dan mengestimasi ketebalan serta kedalaman bidang batas perlapisan bawah permukaan secara sistematis.

---

## 2. Struktur Good Prompt

### R — Role
Anda berperan sebagai seorang **geofisikawan yang ahli dalam metode seismik refraksi dan analisis perlapisan dangkal**.

### T — Task
Tugas Anda adalah mengolah data waktu tiba gelombang pertama untuk menghitung kecepatan tiap lapisan ($V_1, V_2$) dan estimasi kedalaman ($h_1$).

### C — Context
Data diperoleh dari geofon yang membentang pada jarak tertentu dari sumber getaran (*shotpoint*).

### I — Input
1. Nomor geofon[.
2. Jarak geofon dari sumber ($x$).
3. Waktu tiba gelombang pertama ($t$).

### K — Constraints
1. Gunakan Hukum Snellius dan metode *Intercept Time* / *Critical Distance*.
2. Tampilkan rumus dan konversi satuan secara jelas.
3. Tidak merekayasa data.

### O — Output
1. Tabel pengolahan data.
2. Persamaan garis regresi segmen gelombang langsung dan bias.
3. Perhitungan kecepatan ($V_1, V_2$) dan kedalaman ($h_1$).
4. Model perlapisan dan interpretasi litologi.

### Q — Quality Criteria
Perhitungan matematis tepat, menggunakan satuan fisik m/s dan meter, serta interpretasi sesuai kecepatan gelombang.

### V — Verification
Memastikan syarat refraksi kritis terpenuhi ($V_2 > V_1$) dan nilai kedalaman bernilai positif serta logis.

# PROMPT 4 — SEISMIK REFLEKSI

## 1. Narasi Permasalahan
Metode seismik refleksi merupakan salah satu metode geofisika eksplorasi beresolusi tinggi yang digunakan untuk memetakan struktur dan stratigrafi bawah permukaan skala sedang hingga dalam. Metode ini mengandalkan pemantulan gelombang akustik ketika menabrak batas antar medium yang memiliki perbedaan impedansi akustik (perkalian antara kecepatan gelombang dan densitas batuan). Seismik refleksi menjadi standar utama dalam eksplorasi hidrokarbon, cekungan sedimen, serta pemetaan patahan aktif mendalam.

Tantangan utama dalam pengolahan data seismik refleksi adalah mengoreksi efek jarak sumber-penerima (*offset*) terhadap waktu tempuh gelombang pantul, yang dikenal sebagai *Normal Moveout* (NMO). Tanpa koreksi NMO dan estimasi kecepatan RMS ($V_{RMS}$) yang presisi, proses *stacking* (penjumlahan sinyal) tidak akan optimal, sehingga visualisasi penampang seismik menjadi kabur dan sulit untuk mengidentifikasi keberadaan jebakan hidrokarbon maupun deformasi struktur geologi.

---

## 2. Struktur Good Prompt

### R — Role
Anda berperan sebagai seorang  **geofisikawan yang ahli dalam pengolahan data seismik refleksi, analisis impedansi akustik, dan interpretasi struktur sub-surface**.

### T — Task
Tugas Anda adalah mengolah data waktu pantul gelombang (*two-way travel time* / TWT) untuk menentukan *Normal Moveout* (NMO), kecepatan RMS ($V_{RMS}$), dan koefisien refleksi ($R$).

### C — Context
Data seismik refleksi merekam pantulan gelombang dari batas perlapisan bawah permukaan untuk memetakan penampang struktur geologi seperti lipatan atau patahan.

### I — Input
1. Jarak offset sumber ke geofon ($x$).
2. Waktu tempuh dua arah ($t_0$ dan $t_x$).
3. Densitas ($\rho$) dan kecepatan ($V$) tiap lapisan jika tersedia.

### K — Constraints
1. Gunakan persamaan NMO dan koefisien refleksi $R = \frac{\rho_2 V_2 - \rho_1 V_1}{\rho_2 V_2 + \rho_1 V_1}$.
2. Gunakan satuan waktu detik (s) dan kecepatan m/s.
3. Bedakan analisis domain waktu ($t$) dan domain kedalaman ($z$).

### O — Output
1. Tabel kalkulasi koreksi NMO.
2. Nilai koefisien refleksi antar lapisan.
3. Estimasi kedalaman reflektor.
4. Interpretasi struktur bawah permukaan.

### Q — Quality Criteria
Koreksi NMO akurat, koefisien refleksi berada pada rentang $-1$ hingga $1$, dan penjelasannya sistematis.

### V — Verification
Memeriksa apakah waktu $t_x$ selalu lebih besar dari $t_0$ pada offset $x > 0$ dan memastikan nilai NMO *correction* mengecil seiring bertambahnya kedalaman.

# PROMPT 5 — GEOMAGNETIK

## 1. Narasi Permasalahan
Metode geomagnetik merupakan metode geofisika pasif yang memanfaatkan variasi medan magnetik bumi akibat perbedaan nilai susseptibilitas magnetik batuan penyusun bawah permukaan. Metode ini sangat efektif digunakan dalam eksplorasi mineral logam, benda terpendam, delineasi struktur geologi regional seperti sesar, hingga delineasi zona intrusi magmatik. Pengukuran di lapangan merekam medan magnetik total yang merupakan superposisi dari berbagai sumber.

Dalam praktiknya, data medan magnetik hasil pengukuran lapangan tidak dapat langsung diinterpretasikan karena masih terkontaminasi oleh pengaruh medan magnet utama bumi (IGRF) serta variasi harian (*diurnal variation*) akibat aktivitas matahari. Oleh karena itu, seorang geofisikawan harus melakukan reduksi dan koreksi data secara cermat agar diperoleh nilai anomali magnetik murni. Anomali inilah yang kemudian dikorelasikan dengan pola kontur serta kondisi geologi lokal daerah penelitian.

---

## 2. Struktur Good Prompt

### R — Role
Anda berperan sebagai seorang **geofisikawan yang ahli dalam metode geomagnetik dan interpretasi anomali medan magnet bumi**.

### T — Task
Tugas Anda adalah mengolah data medan magnet total, melakukan koreksi harian dan IGRF, serta menentukan pola anomali magnetik.

### C — Context
Data pengukuran medan magnet dipengaruhi oleh medan utama bumi, variasi harian, dan respons material lokal.

### I — Input
1. Koordinat dan waktu pengukura.
2. Medan magnet total ($T_{obs}$).
3. Data variasi harian ($T_{harian}$) dan nilai IGRF ($T_{IGRF}$).

### K — Constraints
1. Gunakan rumus anomali magnetik: $\Delta T = T_{obs} \pm \Delta T_{harian} - T_{IGRF}$.
2. Gunakan satuan nT (nanoTesla).
3. Jangan menentukan jenis batuan secara absolut tanpa data geologi pendukung.

### O — Output
1. Tabel koreksi data magnetik.
2. Nilai anomali magnetik ($\Delta T$).
3. Identifikasi zona anomali tinggi dan rendah.
4. Interpretasi struktur/sifat magnetik dan kesimpulan.

### Q — Quality Criteria
Perhitungan koreksi konsisten, satuan nT tepat, dan interpretasi memperhitungkan ketidakpastian.

### V — Verification
Memeriksa kembali tanda positif/negatif pada koreksi harian dan memastikan nilai anomali magnetik berada pada rentang yang wajar.

## 2. Struktur Komponen Good Prompt (R–T–C–I–K–O–Q–V)

Setiap prompt penyelesaian masalah geofisika di atas dirancang dengan struktur terstandar:

- **R — Role:** Menentukan peran AI sebagai geofisikawan spesialis pada metode terkait.
- **T — Task:** Menjelaskan tugas pengolahan data dan perhitungan secara step-by-step.
- **C — Context:** Memberikan latar belakang fisis dan tujuan pengukuran lapangan.
- **I — Input:** Menyediakan format dan jenis data mentah yang diolah oleh AI.
- **K — Constraints:** Memberikan batasan teknis (rumus baku, satuan, larangan manipulasi data).
- **O — Output:** Menyusun struktur keluaran (tabel, persamaan, grafik, interpretasi, kesimpulan).
- **Q — Quality Criteria:** Menetapkan standar keakuratan fisis dan matematis.
- **V — Verification:** Langkah pengecekan ulang/validasi logis sebelum kesimpulan disajikan.
