```python
# =====================================================================
# RECIPE R-MOD-001: SIMULASI DIFUSI PANAS SATU DIMENSI 
# =====================================================================

def run_detailed_heat_diffusion():
    print("======================================================================")
    print(" MENJALANKAN RECIPE R-MOD-001: SIMULASI DIFUSI PANAS 1D (TERINCI)     ")
    print("======================================================================")
    
    # --- TAHAP 1, 2, & 3: SETUP PARAMETER & KONTRAK INPUT ---
    total_depth = 10.0      # Total kedalaman (meter)
    nz = 50                 # Jumlah titik grid spasial[cite: 20]
    dz = total_depth / (nz - 1)
    alpha = 1e-6            # Difusivitas termal (m^2/s)[cite: 29]
    dt = 3600               # Langkah waktu dalam detik (1 jam)[cite: 11, 20]
    total_steps = 120       # Jumlah iterasi waktu (120 jam / 5 hari)
    
    print(f"[*] Parameter Fisik & Numerik:")
    print(f"    - Kedalaman Total : {total_depth} m")
    print(f"    - Jumlah Grid (Nz): {nz} titik")
    print(f"    - Spasi Grid (dz) : {dz:.4f} m")
    print(f"    - Difusivitas (alpha): {alpha} m^2/s")
    print(f"    - Langkah Waktu (dt): {dt} detik (1 jam)")
    print(f"    - Total Iterasi   : {total_steps} langkah")
    
    # Inisialisasi profil temperatur awal menggunakan list (Suhu awal seragam 10°C)
    T = [10.0 for _ in range(nz)]
    T[0] = 20.0  # Kondisi batas atas: Suhu permukaan tetap 20°C[cite: 29]
    T[-1] = 10.0 # Kondisi batas bawah: Suhu dasar tetap 10°C
    
    # --- TAHAP 4 & 5: IMPLEMENTASI PROCEDURE (Finite-Difference Solver) ---
    print("\n[*] Menjalankan iterasi numerik perambatan panas (Finite-Difference)...")
    
    # Parameter pengecekan stabilitas (Courant number check sederhana)
    cfl = alpha * dt / (dz**2)
    print(f"    - Nilai Parameter Stabilitas (CFL): {cfl:.5f} (Aman jika < 0.5)")
    
    for step in range(1, total_steps + 1):
        T_new = list(T)
        for i in range(1, nz - 1):
            T_new[i] = T[i] + cfl * (T[i+1] - 2 * T[i] + T[i-1])
        # Pertahankan kondisi batas (Dirichlet boundary conditions)
        T_new[0] = 20.0
        T_new[-1] = 10.0
        T = T_new
        
    print("[*] Simulasi numerik selesai.")
    
    # --- ANALISIS HASIL TERINCI ---
    min_temp = min(T)
    max_temp = max(T)
    avg_temp = sum(T) / len(T)
    
    # --- TAHAP 6: MENULIS OUTPUT ARTIFACT (CSV & MARKDOWN TERINCI) ---
    csv_filename = "profile.csv"
    md_filename = "profile-report.md"
    
    # 1. Menulis file CSV data profil
    print(f"[*] Menulis data tabel ke: {csv_filename}")
    with open(csv_filename, "w", encoding="utf-8") as f:
        f.write("depth_m,temperature_c\n")
        for i in range(nz):
            depth = i * dz
            f.write(f"{round(depth, 3)},{round(T[i], 4)}\n")
            
    # 2. Menulis laporan analitis terinci berformat Markdown
    print(f"[*] Menulis laporan terinci ke: {md_filename}")
    report_content = f"""# Laporan Hasil Simulasi: R-MOD-001 (Difusi Panas 1D)

## Ringkasan Eksekusi
- **Status Skenario**: `Executed` (Berhasil)
- **Waktu Simulasi Total**: {total_steps} jam ({total_steps / 24:.1f} hari)
- **Koefisien Difusivitas ($\alpha$)**: `{alpha} m^2/s`
- **Jumlah Titik Kedalaman**: `{nz} titik`

## Statistik Hasil Akhir Suhu Profil
- **Suhu Minimum**: `{min_temp:.4f} °C` (pada kedalaman dasar `{total_depth} m`)
- **Suhu Maksimum**: `{max_temp:.4f} °C` (pada kedalaman permukaan `0.0 m`)
- **Suhu Rata-rata Kolom**: `{avg_temp:.4f} °C`

## Tabel Sampel Data Profil Kedalaman (Sebagian)
| Kedalaman (m) | Temperatur (°C) |
| :--- | :--- |
| {0.0:.3f} | {T[0]:.4f} |
| {round(total_depth * 0.25, 3)} | {T[int(nz*0.25)]:.4f} |
| {round(total_depth * 0.50, 3)} | {T[int(nz*0.50)]:.4f} |
| {round(total_depth * 0.75, 3)} | {T[int(nz*0.75)]:.4f} |
| {total_depth:.3f} | {T[-1]:.4f} |

## Verifikasi & Catatan Pemeriksaan
- Persamaan diferensial konduksi panas berjalan stabil tanpa indikasi ledakan numerik (CFL = `{cfl:.5f}`).
- Seluruh file luaran (`{csv_filename}` dan `{md_filename}`) berhasil digenerasi secara utuh tanpa modifikasi data eksternal yang melanggar kontrak input.
"""

    with open(md_filename, "w", encoding="utf-8") as f:
        f.write(report_content)
        
    # --- TAHAP 7: STATUS & EVIDENCE ---
    print("\n=======================================================================")
    print("[+] [Executed Successfully] Status: Exit code 0.")
    print(f"[+] Output 1 (Data Tabel) tersimpan sebagai : {csv_filename}")
    print(f"[+] Output 2 (Laporan Terinci) tersimpan sebagai: {md_filename}")
    print("=======================================================================")

if __name__ == "__main__":
    run_detailed_heat_diffusion()
```
