## BAB I: PENDAHULUAN

### 1.1 Latar Belakang
Metode geofisika merupakan pendekatan sains kebumian yang menyelidiki struktur, stratigrafi, serta dinamika interior bumi secara tidak langsung (*indirect observation*) melalui pengukuran fenomena fisika di permukaan bumi. Karakteristik bawah permukaan diestimasi melalui respon medan alami maupun medan buatan, seperti perambatan gelombang elastis seismik, resistivitas listrik material, dan variasi medan potensial magnetik bumi.

Dalam praktiknya, data mentah (*raw data*) yang direkam oleh instrumen lapangan tidak dapat langsung mencerminkan kondisi geologi sebenarnya karena masih terkontaminasi oleh faktor geometri bentangan, fluktuasi harian alamiah, pelebaran waktu tempuh, maupun efek interferensi lingkungan. Oleh sebab itu, implementasi algoritma komputasi numerik menggunakan bahasa pemrograman Python menjadi instrumen krusial untuk melakukan reduksi, koreksi, ekstraksi parameter utama, serta interpretasi fisis secara objektif dan akurat.

### 1.2 Tujuan Analisis
1. Menentukan waktu terjadinya gempa (*origin time*, $T_0$) dan rasio kecepatan elastisitas batuan ($V_p/V_s$) melalui metode Diagram Wadati pada cabang seismologi.
2. Menghitung faktor geometri ($K$) dan nilai resistivitas semu ($\rho_a$) konfigurasi Schlumberger untuk memetakan perlapisan batuan dan zona akuifer air tanah.
3. Memisahkan fasa gelombang langsung dan bias kritis untuk mengestimasi kecepatan perlapisan ($V_1, V_2$) serta kedalaman batuan dasar (*bedrock*, $h_1$) menggunakan metode *Intercept Time*.
4. Mengoreksi keterlambatan waktu pantul hiperbolik melalui koreksi *Normal Moveout* (NMO) dan menentukan nilai koefisien refleksi ($R$) pada seismik refleksi.
5. Mereduksi pengaruh variasi harian matahari (*diurnal*) dan medan acuan IGRF guna memperoleh nilai Anomali Magnetik Total ($\Delta T$) untuk delineasi struktur sesar dan zona intrusi magmatik.

## BAB II: LANDASAN TEORI & FORMULASI MATEMATIS

### 2.1 Seismologi — Metode Diagram Wadati
Gelombang gempa bumi merambat melalui interior bumi dalam bentuk gelombang primer/kompresi (gelombang P) dan gelombang sekunder/geser (gelombang S). Dengan asumsi medium homogen isotropik, jarak hiposentral dari sumber gempa ke stasiun ke-$i$ ($D_i$) dirumuskan:
$$D_i = V_p (T_{p,i} - T_0) = V_s (T_{s,i} - T_0)$$
Dengan mengurangkan kedua waktu tempuh, diperoleh hubungan linier Diagram Wadati:
$$(T_{s,i} - T_{p,i}) = \left(\frac{V_p}{V_s} - 1\right) (T_{p,i} - T_0)$$
Bentuk linier OLS: $Y = m \cdot X + c$, di mana:
* Kemiringan (*slope*): $m = \left(\frac{V_p}{V_s} - 1\right) \implies \frac{V_p}{V_s} = m + 1$
* Intersep: $c = -m \cdot T_0 \implies T_0 = -\frac{c}{m}$

### 2.2 Geolistrik Resistivitas — Konfigurasi Schlumberger (VES)
Pada konfigurasi Schlumberger 1D, jarak elektroda arus ($AB$) diperlebar bertahap sedangkan elektroda potensial ($MN$) dipertahankan relatif sempit.
* Faktor Geometri ($K$):
  $$K = \frac{\pi \left[\left(\frac{AB}{2}\right)^2 - \left(\frac{MN}{2}\right)^2\right]}{2 \left(\frac{MN}{2}\right)}$$
* Resistivitas Semu ($\rho_a$):
  $$\rho_a = K \cdot \left(\frac{\Delta V}{I}\right)$$

### 2.3 Seismik Refraksi — Metode Intercept Time (2 Lapisan Datar)
Berdasarkan Hukum Snellius, gelombang bias kritis merambat di sepanjang batas diskontinuitas jika $V_2 > V_1$:
* Gelombang Langsung: $t_{\text{dir}} = \frac{x}{V_1} \implies V_1 = \frac{1}{m_1}$
* Gelombang Bias Kritis: $t_{\text{ref}} = \frac{x}{V_2} + T_i \implies V_2 = \frac{1}{m_2}$
* Kedalaman Batuan Dasar ($h_1$):
  $$h_1 = \frac{T_i \cdot V_1 \cdot V_2}{2 \sqrt{V_2^2 - V_1^2}}$$

### 2.4 Seismik Refleksi — Koreksi NMO & Koefisien Refleksi
Gelombang pantul dari bidang batas horizontal membentuk lintasan hiperbola terhadap offset penerima ($x$):
* Waktu Tempuh Pantul ($t_x$): $t_x = \sqrt{t_0^2 + \left(\frac{x}{V_{\text{RMS}}}\right)^2}$
* Koreksi Normal Moveout: $\Delta t_{\text{NMO}} = t_x - t_0$
* Koefisien Refleksi ($R$):
  $$R = \frac{\rho_2 V_2 - \rho_1 V_1}{\rho_2 V_2 + \rho_1 V_1} = \frac{AI_2 - AI_1}{AI_2 + AI_1}$$

### 2.5 Geomagnetik — Reduksi Medan Total
Nilai medan magnet total hasil pembacaan magnetometer di lapangan ($T_{\text{obs}}$) dikoreksi terhadap variasi harian matahari ($\Delta T_{\text{diurnal}}$) dan medan magnetik global bumi ($T_{\text{IGRF}}$):
$$\Delta T = T_{\text{obs}} - T_{\text{IGRF}} - \Delta T_{\text{diurnal}}$$
